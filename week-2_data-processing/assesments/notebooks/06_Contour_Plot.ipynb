# Contour Plot

A contour plot is a graphical representation of a three-dimensional
surface on a two-dimensional plane.

It uses contour lines to represent points that have the same value.

Contour plots are useful for understanding relationships between two
variables and how a third variable changes across the two-dimensional
space.

## Components of a Contour Plot

A contour plot consists of:

- X-axis: First variable
- Y-axis: Second variable
- Contour lines: Points with equal values
- Color/levels: Represent the magnitude of the third variable

Contour plots are commonly used in data visualization, scientific
analysis, and machine learning.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("../data/WineQT.csv")

df.head()
```

## Creating a Simple Contour Plot

Before applying contour plots to real data, we create a simple mathematical
example.

A contour plot requires:

- X values
- Y values
- A function that generates Z values

The Z values determine the height or intensity represented by contour
lines.

```python
x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)

X, Y = np.meshgrid(x, y)

Z = np.sin(np.sqrt(X**2 + Y**2))

plt.figure(figsize=(8, 6))

plt.contour(X, Y, Z)

plt.title("Simple Contour Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")

plt.show()
```

### Observation

The contour plot represents different levels of the function using contour
lines.

Each line connects points having the same Z-value.

The distance between contour lines indicates the rate of change:
closely spaced lines represent faster changes, while wider spacing
represents slower changes.

## Contour Plot Using Wine Quality Dataset

Contour plots can be used to visualize relationships between numerical
features.

Here we analyze the relationship between:

- Alcohol (X-axis)
- Density (Y-axis)

and visualize the density of observations using contour levels.

```python
x = df["alcohol"]
y = df["density"]

plt.figure(figsize=(8,6))

sns.kdeplot(
    x=x,
    y=y,
    fill=False,
    levels=10
)

plt.title("Contour Plot: Alcohol vs Density")
plt.xlabel("Alcohol")
plt.ylabel("Density")

plt.show()
```

### Observation

The contour plot shows the concentration of observations between alcohol
and density values.

Regions with closely spaced contour lines represent areas with higher
data concentration.

Contour plots help identify patterns and relationships between two
continuous variables.

## Filled Contour Plot

A filled contour plot uses different colors or shades to represent
different value levels.

It provides a better visual understanding of areas with higher or lower
density compared with only contour lines.

```python
plt.figure(figsize=(8,6))

sns.kdeplot(
    x=df["alcohol"],
    y=df["density"],
    fill=True,
    cmap="viridis",
    levels=15
)

plt.title("Filled Contour Plot: Alcohol vs Density")
plt.xlabel("Alcohol")
plt.ylabel("Density")

plt.show()
```

### Observation

The filled contour plot represents the density of observations using
different intensity levels.

The high-density regions indicate where most wine samples are concentrated.

This visualization helps identify patterns between alcohol content and
density that may not be easily visible using individual plots.
