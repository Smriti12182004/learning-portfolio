### Task 2: Selecting Rows and Columns

#### Step 2.1: Select Columns

Select specific columns from the DataFrame for focused analysis.

```python
df["alcohol"]
```

```python
df[["alcohol", "quality"]]
```

#### Step 2.2: Select Rows using `loc[]`.
 Select rows and columns using their labels with `loc[]`.

Filter rows based on conditions applied to DataFrame columns.

### Task 3: Create a New Column

#### Step 3.1: Create Derived Column from Existing Columns

Create a new column using an existing DataFrame column.

```python
df["alcohol_level"] = df["alcohol"] > 12
```

```python
df[["alcohol", "alcohol_level"]].head()
```

#### Step 3.2: Create a Derived Column

Create a new column by combining existing numerical columns.

```python
df["acidity_ratio"] = df["fixed acidity"] / df["volatile acidity"]
```

```python
df[["fixed acidity", "volatile acidity", "acidity_ratio"]].head()
```

### Task 4: Group By and Aggregate

#### Step 4.1: Group by Quality

Group the dataset by the categorical `quality` column.

```python
df.groupby("quality")
```

#### Step 4.2: Calculate Mean and Count

Calculate the mean and count of alcohol content for each quality group.

```python
df.groupby("quality")["alcohol"].agg(["mean", "count"])
```

### Task 5: Merge DataFrames

#### Step 5.1: Create DataFrames

Create two DataFrames with a common key for merging.

```python
df1 = pd.DataFrame({
    "Id": [1, 2, 3, 4],
    "quality": [5, 6, 7, 8]
})

df2 = pd.DataFrame({
    "Id": [1, 2, 3, 4],
    "category": ["Low", "Medium", "Good", "Excellent"]
})
```

#### Step 5.2: Merge DataFrames

Merge the DataFrames using the common `Id` column.

```python
merged_df = pd.merge(df1, df2, on="Id")
merged_df
```

#### Step 5.3: Confirm Resulting Row Count

Verify that the merged DataFrame contains the expected number of rows.

```python
merged_df.shape[0]
```
