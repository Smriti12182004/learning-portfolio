### Task 1: Creating 2D NumPy Array

#### Step 1.1: Import NumPy

Import NumPy for array-based numerical operations.

```python
import numpy as np
```

#### Step 1.2: Create a 2D NumPy Array

Create a 2D array for numerical analysis.

```python
data = np.array([
    [10, 20, 30],
    [15, 25, 35],
    [20, 30, 40],
    [25, 35, 45]
])

data
```

#### Step 1.3 :Compute Row Means and Column Maxima 

Calculate row means and column maxima using NumPy aggregation functions.

```python
row_means = np.mean(data, axis=1)
column_maxima = np.max(data, axis=0)

row_means, column_maxima
```

### Task 2: Normalize Each Column

#### Step 2.1: Normalize Columns

Normalize each column to the range 0–1 using vectorized NumPy operations.

```python
column_min = data.min(axis=0)
column_max = data.max(axis=0)

normalized_data = (data - column_min) / (column_max - column_min)

normalized_data
```

### Task 3: Applying Broadcasting

#### Step 3.1: Subtract Column Means

Use NumPy broadcasting to subtract column means from every row.

```python
column_means = data.mean(axis=0)

centered_data = data - column_means

centered_data
```

### Task 4: Compare Vectorized Operation with Python Loop

#### Step 4.1: Compare Vectorized Operation and Loop

Compare the execution time of a NumPy vectorized operation with an equivalent Python loop.

```python
import time

large_data = np.random.rand(1000000)
```

```python
start = time.time()

vectorized_result = large_data * 2

vectorized_time = time.time() - start

print("Vectorized Time:", vectorized_time)
```

```python
start = time.time()

loop_result = []

for value in large_data:
    loop_result.append(value * 2)

loop_time = time.time() - start

print("Loop Time:", loop_time)
```

```python
print("Speed Difference:", loop_time / vectorized_time)
```

## Overall Observations

- NumPy provides efficient array-based operations for numerical computation.
- Vectorized operations reduce execution time compared to traditional Python loops.
- Broadcasting allows operations between arrays of different shapes without explicit looping.
- Column-wise normalization scales data into a common range (0–1), making it suitable for analysis.
- NumPy aggregation functions such as `mean()` and `max()` simplify statistical computations on arrays.
