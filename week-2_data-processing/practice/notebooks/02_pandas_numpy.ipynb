## What is NumPy?

NumPy (Numerical Python) is a Python library used for numerical
computing.

It provides the `ndarray` object for storing and processing numerical
data efficiently.

NumPy is commonly used for:
- Numerical calculations
- Arrays and matrices
- Mathematical operations
- Statistical operations
- Data processing

```python
arr = np.array([10, 20, 30, 40, 50])

print(arr)
print(type(arr))
print(arr.shape)
print(arr.ndim)
print(arr.dtype)
```

## NumPy Array Operations

NumPy allows us to perform mathematical operations directly on arrays.
These operations are faster and more convenient than performing the same
operations using Python loops.

```python
arr = np.array([10, 20, 30, 40, 50])

print("Addition:", arr + 5)
print("Subtraction:", arr - 5)
print("Multiplication:", arr * 2)
print("Division:", arr / 10)
```

## NumPy Indexing and Slicing

Indexing is used to access individual elements of an array, while slicing
is used to access a range of elements.

```python
print("First element:", arr[0])
print("Last element:", arr[-1])
print("First three elements:", arr[:3])
print("Elements from index 2:", arr[2:])
```

## NumPy 2D Arrays

A 2D NumPy array consists of rows and columns and can be used to
represent tabular or matrix-like data.

The structure of a 2D array can be examined using properties such as
shape, ndim, and size.

```python
data = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

print(data)
print("Shape:", data.shape)
print("Dimensions:", data.ndim)
print("Size:", data.size)
```

## Indexing and Slicing in 2D Arrays

In a 2D array, an element can be accessed using its row and column
index. Python uses zero-based indexing.

```python
print("First element:", data[0, 0])
print("Second row:", data[1])
print("Second column:", data[:, 1])
print("First two rows:")
print(data[:2])
```

## Statistical Operations with NumPy

NumPy provides built-in functions for common statistical calculations
such as mean, median, minimum, maximum, and standard deviation.

```python
values = np.array([10, 20, 30, 40, 50])

print("Mean:", np.mean(values))
print("Median:", np.median(values))
print("Minimum:", np.min(values))
print("Maximum:", np.max(values))
print("Standard deviation:", np.std(values))
```

## Reshaping NumPy Arrays

Reshaping changes the structure of an array without changing its data.

The total number of elements must remain the same before and after
reshaping.

```python
arr = np.arange(1, 13)

print("Original array:")
print(arr)

reshaped = arr.reshape(3, 4)

print("\nReshaped array:")
print(reshaped)

print("\nNew shape:", reshaped.shape)
```

## Conditional Filtering

NumPy allows us to filter array elements using conditions.

This is useful when we need to select observations that satisfy a
specific condition.

```python
values = np.array([10, 25, 40, 15, 60, 30])

filtered = values[values > 30]

print("Original values:", values)
print("Values greater than 30:", filtered)
```

## What is Pandas?

Pandas is a Python library used for data manipulation and analysis.

It provides two important data structures:

- Series: One-dimensional labeled data
- DataFrame: Two-dimensional tabular data

Pandas is commonly used for data cleaning, filtering, transformation,
and analysis.

```python
import pandas as pd

series = pd.Series([10, 20, 30, 40, 50])

print(series)
print("\nType:", type(series))
```

## Pandas Series Indexing

A Series can be accessed using its index.

By default, Pandas assigns integer indexes starting from 0.

```python
print("First value:", series[0])
print("Second value:", series[1])
print("Last value:", series[4])
```

```python
marks = pd.Series(
    [85, 90, 78, 92],
    index=["Maths", "Physics", "Chemistry", "English"]
)

print(marks)
print("\nPhysics marks:", marks["Physics"])
```

## What is a DataFrame?

A DataFrame is a two-dimensional labeled data structure in Pandas.

It is similar to a table with rows and columns.

DataFrames are commonly used for storing, manipulating, and analyzing datasets.

```python
data = {
    "Name": ["Aman", "Riya", "Kartik", "Neha"],
    "Age": [21, 22, 20, 23],
    "Marks": [85, 92, 78, 88]
}

df_demo = pd.DataFrame(data)

print(df_demo)
```

```python
print("Shape:", df_demo.shape)
print("Columns:", df_demo.columns)
print("Data types:")
print(df_demo.dtypes)
```

## Inspecting a DataFrame

Pandas provides useful functions to quickly understand a dataset.

- `head()` shows the first rows.
- `tail()` shows the last rows.
- `info()` shows column names, data types, and non-null values.
- `describe()` gives statistical summary of numerical columns.

```python
df = pd.read_csv("../data/WineQT.csv")

print(df.head())
```

```python
print(df.tail())
```

```python
df.info()
```

```python
df.describe()
```

## Selecting Columns

Pandas allows us to select one or more columns from a DataFrame.

Selecting specific columns is useful when we want to focus on particular features during data analysis.

```python
# Select one column
print(df["alcohol"])
```

```python
# Select multiple columns
selected = df[["alcohol", "pH", "quality"]]

print(selected.head())
```

## Selecting Rows

Rows can be selected using conditions.

This is useful for filtering observations based on specific requirements.

```python
# Wines with quality greater than 6
high_quality = df[df["quality"] > 6]

print(high_quality.head())
```

## Selecting Rows and Columns with loc and iloc

`loc` is used for label-based selection.

`iloc` is used for position-based selection using integer indexes.

```python
# Using loc
print(df.loc[0, "alcohol"])
print(df.loc[0:4, ["alcohol", "pH", "quality"]])
```

```python
# Using iloc
print(df.iloc[0, 10])
print(df.iloc[0:5, 10:13])
```

## Filtering with Multiple Conditions

Pandas allows us to filter rows using multiple conditions.

- `&` means AND
- `|` means OR
- Each condition should be written inside parentheses.

```python
# AND condition
filtered = df[(df["quality"] >= 6) & (df["alcohol"] > 10)]

print(filtered.head())
```

```python
# OR condition
filtered = df[(df["quality"] >= 7) | (df["alcohol"] > 13)]

print(filtered.head())
```

```python
# Multiple columns with conditions
filtered = df[
    (df["quality"] >= 6) &
    (df["pH"] < 3.5) &
    (df["alcohol"] > 10)
]

print(filtered.head())
```

## Sorting DataFrame

Pandas allows us to sort rows based on one or more columns.

`sort_values()` is used to arrange data in ascending or descending order.

```python
# Sort by alcohol in ascending order
sorted_df = df.sort_values("alcohol")

print(sorted_df[["alcohol", "quality"]].head())
```

```python
# Sort by alcohol in descending order
sorted_df = df.sort_values("alcohol", ascending=False)

print(sorted_df[["alcohol", "quality"]].head())
```

## Creating a New Column

A new column can be created by assigning a value or calculation to a new column name.

```python
df["alcohol_percentage"] = df["alcohol"]

print(df[["alcohol", "alcohol_percentage"]].head())
```

## Removing Columns

The `drop()` method is used to remove unwanted rows or columns from a DataFrame.

Removing unnecessary columns is an important part of data cleaning and preprocessing.

```python
# Remove the column we created for practice
df_clean = df.drop(columns=["alcohol_percentage"])

print(df_clean.head())
```

## Creating a Derived Column

A derived column is created from existing columns.

Derived features are commonly used in data processing to represent
information in a more useful form.

```python
df["alcohol_level"] = np.where(
    df["alcohol"] >= 10,
    "High",
    "Low"
)

print(df[["alcohol", "alcohol_level"]].head())
```

## Value Counts

`value_counts()` counts the frequency of each unique value in a column.

It is particularly useful for understanding categorical or discrete variables.

```python
df["quality"].value_counts().sort_index()
```

```python
df["alcohol_level"].value_counts()
```

## Grouping and Aggregation

The `groupby()` method is used to divide data into groups based on the
values of one or more columns.

After grouping, aggregation functions such as `mean()`, `min()`, `max()`,
and `count()` can be used to summarize each group.

This is commonly used in real-world data analysis to compare groups.

```python
df.groupby("quality")["alcohol"].mean()
```

```python
df.groupby("quality")["alcohol"].agg(
    ["count", "mean", "min", "max"]
)
```

```python
df.groupby("quality")[["alcohol", "pH", "density"]].mean()
```

### Named Aggregation

Named aggregation allows us to give meaningful names to calculated
statistics, making the resulting table easier to understand.

```python
quality_summary = df.groupby("quality").agg(
    average_alcohol=("alcohol", "mean"),
    average_pH=("pH", "mean"),
    average_density=("density", "mean"),
    number_of_wines=("quality", "count")
)

quality_summary
```

## Missing Data Handling

Missing values are common in real-world datasets.

Pandas provides methods to identify, count, remove, and replace missing
values.

Common methods include:

- `isnull()` → identifies missing values
- `sum()` → counts missing values
- `dropna()` → removes rows or columns containing missing values
- `fillna()` → replaces missing values

```python
df.isnull().sum()
```

```python
df.isnull().any()
```

```python
sample = pd.Series([10, 20, np.nan, 40, 50])

print("Original:")
print(sample)

print("\nAfter filling:")
print(sample.fillna(sample.mean()))
```

```python
sample = pd.Series([10, 20, np.nan, 40, 50])

print(sample.dropna())
```

### Observation

The WineQT dataset does not contain missing values, so no missing-value
treatment is required for this dataset.

The examples above demonstrate how `fillna()` and `dropna()` can be used
when missing values are present in a real-world dataset.

## Duplicate Data

Duplicate records can occur when the same observation is stored more than
once.

Duplicate data can affect analysis and may lead to incorrect conclusions,
so it is important to identify and handle duplicates during data
processing.

```python
df.duplicated().sum()
```

```python
df[df.duplicated()]
```

```python
df_no_duplicates = df.drop_duplicates()

print("Original rows:", len(df))
print("Rows after removing duplicates:", len(df_no_duplicates))
```

## Renaming Columns

The `rename()` method is used to change column names in a DataFrame.

Clear and meaningful column names make data easier to understand and work with.

```python
df_renamed = df.rename(columns={
    "fixed acidity": "fixed_acidity",
    "volatile acidity": "volatile_acidity",
    "citric acid": "citric_acid",
    "residual sugar": "residual_sugar"
})

print(df_renamed.columns)
```

## Filtering Data with query()

The `query()` method provides a convenient way to filter rows using
conditions written as a string.

```python
high_quality = df.query("quality >= 7")

print(high_quality[["alcohol", "quality"]].head())
```

```python
high_quality = df.query("quality >= 7")

print(high_quality[["alcohol", "quality"]].head())
```

## Practical Data Processing Workflow

A basic data-processing workflow includes:

1. Loading the dataset
2. Inspecting the data
3. Checking missing values and duplicates
4. Filtering required observations
5. Grouping and summarizing data
6. Creating a final summary

```python
# 1. Load dataset
df = pd.read_csv("../data/WineQT.csv")
```

```python
# 2. Inspect dataset
print("Shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())
```

```python
# 3. Check missing values and duplicates
print("\nMissing values:")
print(df.isnull().sum().sum())
```

```python
# 3. Duplicate values
print("\nDuplicate rows:")
print(df.duplicated().sum())
```

```python
# 4. Filter high-quality wines
high_quality = df[df["quality"] >= 7]

print("High-quality wines:", len(high_quality))
print(high_quality[["alcohol", "pH", "quality"]].head())
```

```python
# 5. Group by quality and calculate averages
summary = df.groupby("quality").agg(
    average_alcohol=("alcohol", "mean"),
    average_pH=("pH", "mean"),
    average_density=("density", "mean"),
    number_of_wines=("quality", "count")
)

summary
```
