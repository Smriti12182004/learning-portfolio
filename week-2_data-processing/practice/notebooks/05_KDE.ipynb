# Kernel Density Estimation (KDE)

Kernel Density Estimation (KDE) is a non-parametric method used to
estimate the probability distribution of a dataset.

Unlike histograms, KDE produces a smooth continuous curve that shows
the underlying distribution of data.

KDE is commonly used in Exploratory Data Analysis (EDA) to understand
the shape, spread, and patterns in data.

## KDE vs Histogram

Histogram divides data into fixed intervals (bins) and counts the number
of observations in each interval.

KDE creates a smooth probability density curve without depending on fixed
bins.

Advantages of KDE:

- Shows the overall distribution shape clearly.
- Helps identify peaks and spread.
- Useful for comparing distributions between groups.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("../data/WineQT.csv")

df.head()
```

## KDE Plot of Alcohol Distribution

The alcohol feature is selected to visualize its probability distribution.

KDE curve helps us understand:
- Where most alcohol values are concentrated
- Shape of the distribution
- Presence of skewness

```python
plt.figure(figsize=(8,5))

sns.kdeplot(
    data=df,
    x="alcohol",
    fill=True
)

plt.title("KDE Plot of Alcohol Distribution")
plt.xlabel("Alcohol")
plt.ylabel("Density")

plt.show()
```

### Observation

The KDE plot shows the smooth distribution of alcohol values.

Most wines have alcohol values concentrated around the central region,
while fewer observations are present at extreme values.

The shape of the curve helps identify the spread and skewness of the data.

## Histogram with KDE Curve

Combining a histogram with KDE provides both:

- Frequency distribution of observations
- Smooth estimation of the probability density

This helps in better understanding of data distribution.

```python
plt.figure(figsize=(8,5))

sns.histplot(
    data=df,
    x="alcohol",
    bins=20,
    kde=True
)

plt.title("Histogram with KDE Curve - Alcohol")
plt.xlabel("Alcohol")
plt.ylabel("Frequency")

plt.show()
```

### Observation

The histogram with KDE curve shows the frequency and smooth distribution
of alcohol values.

The peak of the KDE curve represents the range where most observations
are concentrated.

The distribution shows a slight right skew, indicating some wines have
higher alcohol values compared to the majority.

## KDE Bandwidth

Bandwidth is an important parameter in KDE that controls the smoothness
of the density curve.

- Smaller bandwidth → more detailed curve, but may capture noise.
- Larger bandwidth → smoother curve, but may hide important patterns.

Choosing an appropriate bandwidth is important for accurate distribution
analysis.

```python
plt.figure(figsize=(8,5))

sns.kdeplot(
    data=df,
    x="alcohol",
    bw_adjust=0.5,
    label="Bandwidth = 0.5"
)

sns.kdeplot(
    data=df,
    x="alcohol",
    bw_adjust=2,
    label="Bandwidth = 2"
)

plt.title("Effect of Bandwidth on KDE")
plt.xlabel("Alcohol")
plt.ylabel("Density")

plt.legend()
plt.show()
```

### Observation

The bandwidth parameter controls the smoothness of the KDE curve.

A smaller bandwidth produces a more detailed curve that may capture
small variations in the data.

A larger bandwidth produces a smoother curve by reducing small
fluctuations.

The choice of bandwidth depends on the purpose of analysis.

## KDE Comparison Across Wine Quality Groups

KDE can also be used to compare the distribution of a variable across
different groups.

Here we compare alcohol distribution for different wine quality levels
to understand whether quality groups have different patterns.

```python
plt.figure(figsize=(10,6))

for quality in [5, 6, 7]:
    sns.kdeplot(
        data=df[df["quality"] == quality],
        x="alcohol",
        label=f"Quality {quality}"
    )

plt.title("Alcohol Distribution Across Wine Quality Groups")
plt.xlabel("Alcohol")
plt.ylabel("Density")

plt.legend()
plt.show()
```

### Observation

The KDE curves show the distribution of alcohol values for different
wine quality groups.

Higher quality groups generally show a shift towards higher alcohol
values compared with lower quality groups.

The overlap between curves indicates that alcohol alone cannot completely
separate wine quality categories.

## KDE Plot of pH Distribution

KDE can be applied to different numerical features to compare their
distribution patterns.

The pH variable is selected because it showed a relatively symmetric
distribution during EDA.

```python
plt.figure(figsize=(8,5))

sns.kdeplot(
    data=df,
    x="pH",
    fill=True
)

plt.title("KDE Plot of pH Distribution")
plt.xlabel("pH")
plt.ylabel("Density")

plt.show()
```

### Observation

The KDE plot of pH shows a smooth and concentrated distribution.

The curve is approximately symmetric, indicating that pH values are
distributed around a central range with fewer observations at extreme
values.

KDE provides a clearer view of the distribution shape compared with a
simple histogram.
