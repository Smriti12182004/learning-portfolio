### 1. Introduction to Machine Learning

#### What is Machine Learning?

Machine Learning (ML) is a branch of Artificial Intelligence (AI) that allows computers to learn patterns from data and make predictions or decisions without being explicitly programmed.

Instead of manually writing rules for every situation, we provide data to a machine learning algorithm, and the model learns the relationship between inputs and outputs.

### 2. Artificial Intelligence vs Machine Learning


#### Artificial Intelligence (AI)

Artificial Intelligence is a broad field of computer science that focuses on creating systems that can perform tasks which normally require human intelligence.

Examples:

- Decision making
- Problem solving
- Understanding language
- Recognizing images
- Speech recognition


AI is the bigger concept of making machines intelligent.

#### Machine Learning (ML)

Machine Learning is a subset of Artificial Intelligence that allows machines to learn patterns from data and improve their performance without being explicitly programmed.


Instead of writing every rule manually, ML algorithms learn relationships from data.


Examples:

- Spam detection
- House price prediction
- Recommendation systems
- Student score prediction

#### Why is Machine Learning Important?

Traditional programming requires humans to define every rule manually.

However, many real-world problems are complex because:

- Data is very large
- Patterns are difficult to identify manually
- Rules may change over time


Machine Learning helps computers automatically discover patterns from data and make predictions.

### 3. Why Do We Need Machine Learning?


Traditional programming works when the rules are simple and clearly defined.

Example:

A programmer can create rules:

- If marks >= 40 → Pass
- If marks < 40 → Fail


But many real-world problems cannot be solved by writing fixed rules because:

- The amount of data is very large.
- Patterns are complex.
- It is difficult for humans to define every possible condition.


Machine Learning solves this problem by learning patterns from examples instead of depending on manually written rules.

#### Traditional Rules vs Machine Learning

#### Machine Learning Approach

The model learns rules from data.

Flow:

Example:

#### Key Question


#### Why do we use Machine Learning instead of writing rules manually?


For simple problems, manually written rules work.

However, for complex problems like image recognition, speech recognition, and recommendation systems, defining every rule is difficult.

Machine Learning allows computers to automatically learn patterns from data.

### 4. Traditional Programming vs Machine Learning


#### Traditional Programming

In traditional programming, humans define the rules and logic.

The computer follows the instructions written by the programmer.


Flow:

Rules + Data → Output


Example:

Rules:

- If marks < 40 → Fail

Flow:

- Previous Score


Target:
- Final Score

### 5. Types of Machine Learning


Machine Learning is mainly divided into three categories:

1. Supervised Learning
2. Unsupervised Learning
3. Reinforcement Learning

#### 5.1 Supervised Learning


Supervised Learning is a type of Machine Learning where the model learns from labelled data.


In labelled data:

- Input data (Features) is provided.
- Correct output (Target) is provided.


Flow:


Features (X) + Target (y)

↓

Machine Learning Model

↓

Prediction


Example:

Student Performance Dataset:


Features:

- Hours Studied
- Attendance
- Previous Score


Target:

- Final Score


The model learns the relationship between features and target and predicts outputs for new data.

#### Types of Supervised Learning


Supervised Learning is mainly divided into:


#### 1. Regression

Regression is used when the output is a continuous numerical value.


Examples:

- House price prediction
- Salary prediction
- Student marks prediction


Output example:

85.5
₹50,000
₹75,00,000


The output is a number.

#### 2. Classification


Classification is used when the output belongs to a category or class.


Examples:

- Spam / Not Spam
- Pass / Fail
- Disease / No Disease


Output example:

0 or 1

Yes or No

Category labels


The output represents a class.

#### 5.2 Unsupervised Learning


Unsupervised Learning is a type of Machine Learning where the model learns patterns from data without labelled outputs.


In this type of learning:

- Only input data is provided.
- The model identifies hidden patterns or structures in the data.


Flow:


Input Data (X)

↓

Machine Learning Model

↓

Find Patterns


Examples:

- Customer segmentation
- Grouping similar products
- Finding hidden patterns in datasets

#### 5.3 Reinforcement Learning


Reinforcement Learning is a type of Machine Learning where an agent learns by interacting with an environment.


The model learns through:

- Actions
- Rewards
- Penalties


The goal is to choose actions that maximize rewards.


Learning process:


Environment

↓

Agent takes action

↓

Receives reward or penalty

↓

Improves future decisions


Examples:

- Game-playing AI
- Robotics
- Self-learning systems

#### Comparison of Machine Learning Types


| Type | Data Required | Learning Method | Example |
|---|---|---|---|
| Supervised Learning | Labelled data (Input + Output) | Learns relationship between features and target | House price prediction |
| Unsupervised Learning | Unlabelled data | Finds patterns and groups | Customer segmentation |
| Reinforcement Learning | Environment + Rewards | Learns through actions and feedback | Game AI |

### 7. How Does a Machine Learning Model Learn?


A machine learning model learns by identifying patterns in data and improving its predictions through repeated adjustments.


The basic learning process:


Data

↓

Machine Learning Model

↓

Prediction

↓

Compare Prediction with Actual Output

↓

Calculate Error

↓

Calculate Loss

↓

Optimize Parameters

↓

Update Model

↓

Better Predictions


The goal of training is to reduce errors and improve the model's performance.

#### Error


Error represents the difference between the actual output and the predicted output.


Formula:


Error = Actual Value - Predicted Value


Example:


Actual Score = 90

Predicted Score = 85


Error:

90 - 85 = 5


Error tells us how far the model's prediction is from the actual value.

#### Loss


Loss is a value that measures how wrong the model's prediction is.


It converts the error into a measurable value that helps the model understand its performance.


Example:

If predictions are far from actual values:

→ High Loss


If predictions are close to actual values:

→ Low Loss


The goal of a machine learning model is to minimize loss.

#### Optimization


Optimization is the process of improving the model by reducing the loss.


The model changes its parameters to find better values that produce more accurate predictions.


Example:

In Linear Regression:


y = wx + b


The model learns better values of:

- Weight (w)
- Bias (b)


to reduce prediction error.

### Important Questions


#### What happens when we train a model?


During training:

- The model learns patterns from data.
- It makes predictions.
- It calculates errors.
- It updates parameters to improve performance.


#### Why do we calculate error?


Error tells the model how different its prediction is from the actual output.


#### Why do we minimize loss?


Lower loss means predictions are closer to actual values, which means the model performs better.
